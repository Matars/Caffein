```python
import cdsapi

dataset = "cams-global-greenhouse-gas-inversion"
request = {
    "variable": "methane",
    "quantity": "concentration",
    "input_observations": "surface_satellite",
    "time_aggregation": "daily_mean",
    "version": "latest",
    "year": ["2023"],
    "month": ["01"]
}

client = cdsapi.Client(api_url="https://cds.climate.copernicus.eu/api/v2",
                       api_key="your_api_key")
client.retrieve(dataset, request).download()
```
