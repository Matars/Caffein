```python
import pandas as pd
from pathlib import Path

path_volc = Path("../data/kaggle_raw/volcano-events.csv")
# check spelling of 'occurence'
path_fire = Path("../data/kaggle_raw/fire-occurence.csv")


def load_csv(p: Path, name: str):
    if not p.exists():
        raise FileNotFoundError(f"{name} file not found at {p.resolve()}")
    df = pd.read_csv(p)
    print(f"{name}: {df.shape[0]} rows, {df.shape[1]} cols")
    return df


volcano_df = load_csv(path_volc, "Volcano")
fire_df = load_csv(path_fire, "Fire")

print("\nVolcano columns sample:", list(volcano_df.columns[:8]))
print("Fire columns sample:", list(fire_df.columns[:8]))

# If you want a single long dataframe (only for columns in common):
common_cols = list(set(volcano_df.columns) & set(fire_df.columns))
if common_cols:
    combined_df = pd.concat(
        [
            volcano_df[common_cols].assign(_source="volcano"),
            fire_df[common_cols].assign(_source="fire")
        ],
        ignore_index=True
    )
    print(f"\nCombined (common columns) shape: {combined_df.shape}")
else:
    print("\nNo common columns to combine directly.")

# Display heads
print("\nVolcano head:")
display(volcano_df.head())
print("\nFire head:")
display(fire_df.head())
```
